```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Sales"], bins=30)

plt.title("Sales Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Profit"], bins=30)

plt.title("Profit Distribution")

plt.show()
```

```python
sales_category = df.groupby("Category")["Sales"].sum().sort_values()

sales_category.plot(kind="barh", figsize=(8,5))

plt.title("Sales by Category")

plt.xlabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=df,
    x="Sales",
    y="Sub-Category",
    estimator=sum
)

plt.title("Sales by Sub-Category")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind="pie", autopct="%1.1f%%")

plt.ylabel("")

plt.title("Sales by Region")

plt.show()
```

```python
sns.countplot(data=df, x="Segment")

plt.title("Customer Segment Distribution")

plt.show()
```

```python
payment = df.groupby("Payment Mode")["Sales"].sum()

payment.plot(kind="pie", autopct="%1.1f%%")

plt.ylabel("")

plt.title("Sales by Payment Mode")

plt.show()
```

```python
top_states = (
    df.groupby("State")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_states.plot(kind="bar")

plt.title("Top 10 States by Sales")

plt.show()
```

```python
top_customer = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customer.plot(kind="barh")

plt.title("Top Customers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Quantity"], bins=15)

plt.title("Quantity Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit"
)

plt.title("Sales vs Profit")

plt.show()
```

```python
ship = df.groupby("Ship Mode")["Sales"].sum()

ship.plot(kind="bar")

plt.title("Sales by Ship Mode")

plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products.plot(kind="barh")

plt.title("Top Selling Products")

plt.show()
```

```python
print(f"Total Sales : {df['Sales'].sum():,.2f}")

print(f"Total Profit : {df['Profit'].sum():,.2f}")

print(f"Average Sales : {df['Sales'].mean():,.2f}")

print(f"Average Profit : {df['Profit'].mean():,.2f}")

print(f"Unique Customers : {df['Customer ID'].nunique()}")

print(f"Total Orders : {df['Order ID'].nunique()}")
```
